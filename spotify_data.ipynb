```python
import pandas as pd

url = "https://raw.githubusercontent.com/rfordatascience/tidytuesday/master/data/2020/2020-01-21/spotify_songs.csv"
df = pd.read_csv(url)
```

```python
print(df.shape)
print(df.columns.tolist())
```

```python
df.head()
```

```python
print(df['track_popularity'].describe())
print(df['playlist_genre'].value_counts())
```

```python
df['track_popularity'].plot.hist(bins=30)
```

```python
import matplotlib.pyplot as plt
df['track_popularity'].hist(bins=30)
plt.title('Distribución de popularidad')
plt.show()
```

```python
print(df[df['track_popularity'] == 0].shape[0])
```

```python
print(df_0.describe())
```

```python
df_clean = df[df['track_popularity'] > 0].copy()
print(df_clean.shape)
```

```python
df_clean['track_popularity'].hist(bins=30)
plt.title('Distribución de popularidad')
plt.show()
```

```python
df_clean['track_popularity'].describe()
```

```python
import seaborn as sns

correlaciones = df_clean[['track_popularity','danceability','energy',
                           'loudness','speechiness','acousticness',
                           'instrumentalness','liveness','valence','tempo']].corr()

sns.heatmap(correlaciones, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlaciones con popularidad')
plt.show()
```

```python
print(df_clean.groupby('playlist_genre')['track_popularity'].mean().sort_values(ascending=False))
```
